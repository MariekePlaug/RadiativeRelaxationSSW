```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import netCDF4 as netcdf4
```

```python
data_2004 = xr.open_dataset("../data/2004_data.nc", engine="netcdf4")
data_2006 = xr.open_dataset("../data/2006_data.nc", engine="netcdf4")
data_2007 = xr.open_dataset("../data/2007_data.nc", engine="netcdf4")
data_2008 = xr.open_dataset("../data/2008_data.nc", engine="netcdf4")
data_2009 = xr.open_dataset("../data/2009_data.nc", engine="netcdf4")
data_2010 = xr.open_dataset("../data/2010_data.nc", engine="netcdf4")
data_2013 = xr.open_dataset("../data/2013_data.nc", engine="netcdf4")
data_2013
```

```python
fig, ax = plt.subplots(1, figsize=(8,12))

ax.plot(data_2007.t[0,:], data_2007.pressure, label="t = 0")
ax.plot(data_2007.t[1,:], data_2007.pressure, label="t = 1")
ax.plot(data_2007.t[2,:], data_2007.pressure, label="t = 2")
ax.plot(data_2007.t[3,:], data_2007.pressure, label="t = 3")
ax.plot(data_2007.t[4,:], data_2007.pressure, label="t = 4")
ax.plot(data_2007.t[5,:], data_2007.pressure, label="t = 5")
# ax.plot(data_2007.t[12,:], data_2007.pressure, label="t = 12")
# ax.plot(data_2007.t[14,:], data_2007.pressure, label="t = 14")
ax.invert_yaxis()
ax.set_yscale("log")
plt.legend()
```
